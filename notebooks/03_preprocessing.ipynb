# 03 - Preprocessing
Normalize pixel values and create training/validation arrays.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np

train = pd.read_csv("../data/raw/mnist_train.csv")
X = train.drop(columns=["label"]).values.astype("float32") / 255.0
y = train["label"].values
X = X.reshape(-1, 28, 28, 1)

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.1, random_state=42, stratify=y
)
print(X_train.shape, X_val.shape)
```

Pixel values are scaled from 0–255 to 0–1, then reshaped into 28×28×1 images.
